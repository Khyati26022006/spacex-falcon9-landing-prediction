# EDA with Data Visualization

Scatter plots, a bar chart and a line chart explore how site, orbit, payload and time relate to landing success (green = landed, red = failed).

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style="whitegrid")
d=pd.read_csv("data/spacex_wrangled.csv")
dash=pd.read_csv("data/spacex_launch_dash.csv")
pal={0:"#D64545",1:"#1E9E5A"}
```

```python
sns.catplot(data=dash,x="Flight Number",y="Launch Site",hue="class",palette=pal,aspect=2.4); plt.title("Flight number vs launch site"); plt.show()
```

```python
sns.catplot(data=dash,x="Payload Mass (kg)",y="Launch Site",hue="class",palette=pal,aspect=2.4); plt.title("Payload vs launch site"); plt.show()
```

```python
o=d.groupby("Orbit")["Class"].agg(["mean","count"]).sort_values("mean",ascending=False)
o["mean"].plot.bar(color="#0B49CB",figsize=(9,4)); plt.ylabel("Success rate"); plt.title("Success rate by orbit"); plt.show(); o
```

```python
sns.catplot(data=d,x="FlightNumber",y="Orbit",hue="Class",palette=pal,aspect=2.4); plt.title("Flight number vs orbit"); plt.show()
```

```python
sns.catplot(data=d,x="PAYLOAD_MASS__KG_",y="Orbit",hue="Class",palette=pal,aspect=2.4); plt.title("Payload vs orbit"); plt.show()
```

```python
y=d.groupby(d.Date.str[:4])["Class"].mean()
y.plot(marker="o",color="#0B49CB",figsize=(9,4)); plt.ylabel("Success rate"); plt.xlabel("Year"); plt.title("Yearly landing success"); plt.show(); y.round(3)
```
